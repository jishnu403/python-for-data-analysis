```python
import pandas as pd

url = "https://raw.githubusercontent.com/GeethaGunasekaran1/Dataset_rep/main/social_media_engagement_5000.csv"
df = pd.read_csv(url)
df.head()
```

```python
print(" First 5 Rows ")
display(df.head())
```

```python
print("\n Data Types ")
print(df.dtypes)
```

```python
if 'date' in df.columns:
    df['date'] = pd.to_datetime(df['date'])
    print("\nDate column converted to datetime successfully.")
```

```python
import numpy as np
```

```python
if 'likes' in df.columns:
    likes_array = np.array(df['likes'])
    
    print("\n NumPy Array Operations on 'likes' ")
    print("Mean Likes:", np.mean(likes_array))
    print("Median Likes:", np.median(likes_array))
    print("Max Likes:", np.max(likes_array))
    print("Min Likes:", np.min(likes_array))
```

```python
print(df.columns)
```

```python
for col in df.select_dtypes(include=[np.number]).columns:
    df[col] = df[col].fillna(df[col].median())

for col in df.select_dtypes(include=['object']).columns:
    mode_val = df[col].mode()[0] if not df[col].mode().empty else "Unknown"
    df[col] = df[col].fillna(mode_val)

print("Duplicates before cleaning:", df.duplicated().sum())
df = df.drop_duplicates()
print("Duplicates after cleaning:", df.duplicated().sum())

metrics = ['likes', 'comments', 'shares'] 
for col in metrics:
    if col in df.columns:
        df[col] = df[col].apply

if 'gender' in df.columns:
    df['gender'] = df['gender'].str.lower().str.strip()

if 'hashtags' in df.columns:
    df['hashtag_count'] = df['hashtags'].astype(str).apply
if 'sentiment' in df.columns:
    df['sentiment'] = df['sentiment'].str.lower().str.strip()

print("Done")
display(df.head())
```

```python
display(df.head())
display(df.tail())
print(df.shape)
print(df.columns.tolist())
print(df.dtypes)
df.info()
display(df.describe())

cat_col = 'post_type' if 'post_type' in df.columns else df.select_dtypes(include=['object']).columns[0]
display(df[cat_col].value_counts())
print(df[cat_col].unique())
print(df[cat_col].nunique())

numeric_df = df.select_dtypes(include=['number'])
corr_matrix = numeric_df.corr()
display(corr_matrix)

df['likes'] = pd.to_numeric(df['likes'], errors='coerce')
if 'impressions' in df.columns:
    df['impressions'] = pd.to_numeric(df['impressions'], errors='coerce')

if 'post_type' in df.columns and 'likes' in df.columns:
    display(df.groupby('post_type')['likes'].mean())

if 'country' in df.columns and 'impressions' in df.columns:
    display(df.groupby('country')['impressions'].mean())
```

```python
df['likes'] = pd.to_numeric(df['likes'], errors='coerce')
df['comments'] = pd.to_numeric(df['comments'], errors='coerce')
df['shares'] = pd.to_numeric(df['shares'], errors='coerce')

df['engagement_score'] = df['likes'] + (df['comments'] * 2) + (df['shares'] * 3)
df['log_likes'] = np.log1p(df['likes'])

display(df.groupby('post_type')[['engagement_score', 'likes', 'comments', 'shares']].mean())
display(df.groupby('country')[['engagement_score', 'impression_count', 'follower_count']].mean())
display(df.groupby('sentiment')[['engagement_score', 'likes', 'engagement_rate']].mean())

print("Done")
```

```python
columns_to_analyze = ['likes', 'comments', 'shares', 'watch_time_sec', 'engagement_rate', 'follower_count']

for col in columns_to_analyze:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')
stats_dict = {}
for col in columns_to_analyze:
    if col in df.columns:
        mean_val = df[col].mean()
        median_val = df[col].median()
        mode_val = df[col].mode().tolist()
        std_val = df[col].std()
        var_val = df[col].var()
        percentiles = df[col].quantile([0.25, 0.50, 0.75]).to_dict()
        
        stats_dict[col] = {
            'Mean': mean_val,
            'Median': median_val,
            'Mode': mode_val,
            'Std Dev': std_val,
            'Variance': var_val,
            'Percentiles (25, 50, 75)': percentiles
        }
stats_df = pd.DataFrame(stats_dict)
display(stats_df)
print("Done")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

for col in ['likes', 'impression_count', 'age', 'engagement_rate', 'follower_count', 'watch_time_sec']:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

if 'posted_at' in df.columns:
    df['posted_at'] = pd.to_datetime(df['posted_at'], dayfirst=True, errors='coerce')

plt.figure(figsize=(8, 5))
plt.scatter(df['impression_count'], df['likes'], alpha=0.5, color='blue')
plt.title('Likes vs Impressions')
plt.xlabel('Impressions')
plt.ylabel('Likes')
plt.show()

if 'posted_at' in df.columns:
    plt.figure(figsize=(10, 4))
    daily_trend = df.groupby(df['posted_at'].dt.date)['engagement_rate'].mean()
    plt.plot(daily_trend.index, daily_trend.values, marker='o', color='green')
    plt.title('Daily Engagement Trend')
    plt.xlabel('Date')
    plt.ylabel('Avg Engagement Rate')
    plt.xticks(rotation=45)
    plt.show()

if 'post_category' in df.columns:
    plt.figure(figsize=(8, 4))
    df['post_category'].value_counts().plot(kind='bar', color='orange')
    plt.title('Posts by Category')
    plt.xlabel('Category')
    plt.ylabel('Count')
    plt.show()

if 'gender' in df.columns:
    plt.figure(figsize=(6, 6))
    df['gender'].value_counts().plot(kind='pie', autopct='%1.1f%%', startangle=140)
    plt.title('Gender Distribution')
    plt.ylabel('')
    plt.show()

if 'age' in df.columns:
    plt.figure(figsize=(8, 4))
    plt.hist(df['age'].dropna(), bins=20, color='purple', edgecolor='black')
    plt.title('Age Distribution')
    plt.xlabel('Age')
    plt.ylabel('Frequency')
    plt.show()

if 'engagement_rate' in df.columns:
    plt.figure(figsize=(6, 4))
    plt.boxplot(df['engagement_rate'].dropna(), vert=False)
    plt.title('Box Plot of Engagement Rate')
    plt.xlabel('Engagement Rate')
    plt.show()

if 'post_type' in df.columns:
    plt.figure(figsize=(8, 4))
    sns.countplot(data=df, x='post_type', hue='post_type', palette='Set2', legend=False)
    plt.title('Count Plot of Post Type')
    plt.show()

if 'post_category' in df.columns and 'likes' in df.columns:
    plt.figure(figsize=(10, 4))
    sns.barplot(data=df, x='post_category', y='likes', hue='post_category', estimator=np.mean, palette='viridis', legend=False)
    plt.title('Average Likes by Category')
    plt.xticks(rotation=45)
    plt.show()

if 'sentiment' in df.columns and 'follower_count' in df.columns:
    plt.figure(figsize=(8, 4))
    sns.violinplot(data=df, x='sentiment', y='follower_count', hue='sentiment', palette='muted', legend=False)
    plt.title('Followers vs Sentiment')
    plt.show()

numeric_cols = df.select_dtypes(include=['number']).columns[:4]
sns.pairplot(df[numeric_cols].dropna())
plt.show()

plt.figure(figsize=(8, 6))
sns.heatmap(df.select_dtypes(include=['number']).corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix Heatmap')
plt.show()

if 'device_type' in df.columns and 'engagement_rate' in df.columns:
    plt.figure(figsize=(10, 4))
    sns.swarmplot(data=df.sample(min(len(df), 500)), x='device_type', y='engagement_rate', hue='device_type', palette='Set1', legend=False)
    plt.title('Engagement Rate vs Device Type')
    plt.show()

if 'post_type' in df.columns and 'engagement_rate' in df.columns:
    display(df.groupby('post_type')['engagement_rate'].mean())

if 'post_category' in df.columns and 'likes' in df.columns:
    display(df.groupby('post_category')['likes'].mean())

if 'country' in df.columns and 'engagement_rate' in df.columns:
    display(df.groupby('country')['engagement_rate'].mean())

if 'is_verified' in df.columns:
    display(df.groupby('is_verified')[['likes', 'engagement_rate']].mean())

if 'device_type' in df.columns and 'watch_time_sec' in df.columns:
    display(df.groupby('device_type')['watch_time_sec'].mean())

if 'sentiment' in df.columns:
    display(df.groupby('sentiment')[['likes', 'engagement_rate']].mean())

print("Done") 
```
